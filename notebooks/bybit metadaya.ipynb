# SPOT

```python
import requests
from pprint import pprint

URL = "https://api.bybit.com/v5/market/instruments-info"

response = requests.get(
    URL,
    params={"category": "spot"},
    timeout=30,
)

response.raise_for_status()
data = response.json()

print("COUNT:", len(data["result"]["list"]))
print()

pprint(data["result"]["list"][0])
```

```python
btc = next(
    x
    for x in data["result"]["list"]
    if x["symbol"] == "BTCUSDT"
)

pprint(btc)
```

# LINEAR

```python
import requests
from pprint import pprint

URL = "https://api.bybit.com/v5/market/instruments-info"

response = requests.get(
    URL,
    params={
        "category": "linear",
        "symbol": "BTCUSDT",
    },
    timeout=30,
)

response.raise_for_status()
data = response.json()

print("RET CODE:", data["retCode"])
print("COUNT:", len(data["result"]["list"]))
print()

pprint(data["result"]["list"][0])
```

```python
pprint({
    "category": data["result"].get("category"),
    "nextPageCursor": data["result"].get("nextPageCursor"),
})
```

```python
response = requests.get(
    URL,
    params={
        "category": "linear",
        "limit": 1000,
    },
    timeout=30,
)

response.raise_for_status()
data = response.json()

print("COUNT:", len(data["result"]["list"]))
print("NEXT CURSOR:", data["result"].get("nextPageCursor"))
```

# INVERSE

```python
import requests
from pprint import pprint

URL = "https://api.bybit.com/v5/market/instruments-info"

response = requests.get(
    URL,
    params={
        "category": "inverse",
        "symbol": "BTCUSD",
    },
    timeout=30,
)

response.raise_for_status()
data = response.json()

print("RET CODE:", data["retCode"])
print("COUNT:", len(data["result"]["list"]))
print()

pprint(data["result"]["list"][0])
```

```python
response = requests.get(
    URL,
    params={
        "category": "inverse",
        "limit": 1000,
    },
    timeout=30,
)

response.raise_for_status()
data = response.json()

print("COUNT:", len(data["result"]["list"]))
print("NEXT CURSOR:", data["result"].get("nextPageCursor"))

print()
print("CONTRACT TYPES:")

for contract_type in sorted(
    {x["contractType"] for x in data["result"]["list"]}
):
    print(contract_type)
```

# OPTION

```python
import requests
from pprint import pprint

URL = "https://api.bybit.com/v5/market/instruments-info"

response = requests.get(
    URL,
    params={
        "category": "option",
        "baseCoin": "BTC",
        "limit": 1000,
    },
    timeout=30,
)

response.raise_for_status()
data = response.json()

print("RET CODE:", data["retCode"])
print("COUNT:", len(data["result"]["list"]))
print("NEXT CURSOR:", data["result"].get("nextPageCursor"))
print()

pprint(data["result"]["list"][0])
```

```python
options = data["result"]["list"]

print("OPTION TYPES:")
for value in sorted({x["optionsType"] for x in options}):
    print(value)

print()
print("BASE COINS:")
for value in sorted({x["baseCoin"] for x in options}):
    print(value)

print()
print("QUOTE COINS:")
for value in sorted({x["quoteCoin"] for x in options}):
    print(value)

print()
print("SETTLE COINS:")
for value in sorted({x["settleCoin"] for x in options}):
    print(value)

print()
print("FIELDS:")
for field in sorted(set().union(*(x.keys() for x in options))):
    print(field)

print()
print("FIRST 10:")
for x in options[:10]:
    print(
        x["symbol"],
        "| type =", x["optionsType"],
        "| delivery =", x["deliveryTime"],
        "| base =", x["baseCoin"],
        "| quote =", x["quoteCoin"],
        "| settle =", x["settleCoin"],
    )
```
