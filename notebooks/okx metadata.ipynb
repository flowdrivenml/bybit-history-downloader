```python
import requests
from pprint import pprint

URL = "https://www.okx.com/api/v5/public/instruments"

response = requests.get(
    URL,
    params={
        "instType": "SPOT",
    },
    timeout=30,
)

response.raise_for_status()
data = response.json()

print("CODE:", data["code"])
print("COUNT:", len(data["data"]))
print()

btc = next(
    x
    for x in data["data"]
    if x["instId"] == "BTC-USDT"
)

pprint(btc)
```

```python
instruments = data["data"]

print("FIELDS:")
for field in sorted(
    set().union(*(x.keys() for x in instruments))
):
    print(field)
```

```python
print()
print("FIRST 10:")

for x in instruments[:10]:
    print(
        x["instId"],
        "| base =", x.get("baseCcy"),
        "| quote =", x.get("quoteCcy"),
        "| settle =", x.get("settleCcy"),
        "| state =", x.get("state"),
    )
```

# SWAP

```python
import requests
from pprint import pprint

URL = "https://www.okx.com/api/v5/public/instruments"

response = requests.get(
    URL,
    params={
        "instType": "SWAP",
    },
    timeout=30,
)

response.raise_for_status()
data = response.json()

print("CODE:", data["code"])
print("COUNT:", len(data["data"]))
```

```python
instruments = data["data"]

print("CT TYPES:")
for value in sorted({x["ctType"] for x in instruments}):
    print(value)

print()
print("CONTRACT VALUE CURRENCIES:")
for value in sorted({x["ctValCcy"] for x in instruments}):
    print(repr(value))
```

```python
linear = next(
    x
    for x in instruments
    if x["instId"] == "BTC-USDT-SWAP"
)

pprint(linear)
```

```python
for key in [
    "instId",
    "instType",
    "ctType",
    "uly",
    "instFamily",
    "settleCcy",
    "ctVal",
    "ctMult",
    "ctValCcy",
    "tickSz",
    "lotSz",
    "minSz",
    "listTime",
    "expTime",
    "state",
]:
    print(f"{key}: {linear.get(key)!r}")
```

```python
inverse = next(
    x
    for x in instruments
    if x["instId"] == "BTC-USD-SWAP"
)

pprint(inverse)
```

```python
for key in [
    "instId",
    "instType",
    "ctType",
    "uly",
    "instFamily",
    "baseCcy",
    "quoteCcy",
    "settleCcy",
    "ctVal",
    "ctMult",
    "ctValCcy",
    "tickSz",
    "lotSz",
    "minSz",
    "listTime",
    "expTime",
    "state",
]:
    print(f"{key}: {inverse.get(key)!r}")
```

# FUTURE

```python
linear_future = next(
    x
    for x in instruments
    if x["instType"] == "FUTURES"
    and x["ctType"] == "linear"
    and x["instFamily"] == "BTC-USD_UM"
)

inverse_future = next(
    x
    for x in instruments
    if x["instType"] == "FUTURES"
    and x["ctType"] == "inverse"
    and x["instFamily"] == "BTC-USD"
)

print("=== LINEAR FUTURE ===")
pprint(linear_future)

print()
print("=== INVERSE FUTURE ===")
pprint(inverse_future)
```

```python
FIELDS = [
    "instId",
    "instType",
    "ctType",
    "uly",
    "instFamily",
    "baseCcy",
    "quoteCcy",
    "settleCcy",
    "ctVal",
    "ctMult",
    "ctValCcy",
    "tickSz",
    "lotSz",
    "minSz",
    "listTime",
    "expTime",
    "alias",
    "state",
]

print("=== LINEAR FUTURE ===")
for key in FIELDS:
    print(f"{key}: {linear_future.get(key)!r}")

print()
print("=== INVERSE FUTURE ===")
for key in FIELDS:
    print(f"{key}: {inverse_future.get(key)!r}")
```

```python
import requests

URL = "https://www.okx.com/api/v5/public/instruments"

families = [
    "SOL-USD",
    "BTC-USD",
    "ETH-USD",
    "XAU-USD",
]

for family in families:
    response = requests.get(
        URL,
        params={
            "instType": "OPTION",
            "instFamily": family,
        },
        timeout=30,
    )

    print(
        family,
        "status =", response.status_code,
    )

    if response.ok:
        data = response.json()
        print(
            "  code =", data["code"],
            "count =", len(data["data"]),
        )
    else:
        print(" ", response.text[:500])
```
