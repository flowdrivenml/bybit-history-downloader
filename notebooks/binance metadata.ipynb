```python
import requests
from pprint import pprint

URL = "https://dapi.binance.com/dapi/v1/exchangeInfo"

response = requests.get(
    URL,
    timeout=30,
)

response.raise_for_status()
data = response.json()

print("COUNT:", len(data["symbols"]))

btc = next(
    x
    for x in data["symbols"]
    if x["symbol"] == "BTCUSD_PERP"
)

pprint(btc)
```

```python
symbols = data["symbols"]

print("FIELDS:")

for field in sorted(
    set().union(*(x.keys() for x in symbols))
):
    print(field)
```

```python
FIELDS = [
    "symbol",
    "pair",
    "contractType",
    "contractSize",
    "deliveryDate",
    "onboardDate",
    "status",
    "baseAsset",
    "quoteAsset",
    "marginAsset",
    "pricePrecision",
    "quantityPrecision",
    "underlyingType",
]

for key in FIELDS:
    print(f"{key}: {btc.get(key)!r}")

print()
print("FILTERS:")

for f in btc["filters"]:
    pprint(f)
```

```python
print("CONTRACT TYPES:")

for value in sorted(
    {x["contractType"] for x in symbols}
):
    print(repr(value))
```
