```python
import requests
from pprint import pprint

URL = "https://api.binance.com/api/v3/exchangeInfo"

response = requests.get(
    URL,
    timeout=30,
)

response.raise_for_status()
data = response.json()

print("COUNT:", len(data["symbols"]))

btc = next(
    x
    for x in data["symbols"]
    if x["symbol"] == "BTCUSDT"
)

pprint(btc)
```
